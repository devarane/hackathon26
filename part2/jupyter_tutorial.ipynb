## 1. What is a Jupyter notebook?

A notebook is a document that mixes **markdown** (notes, headings, explanations) with **code cells** that you can run one at a time. The results stay in the notebook, which makes it a good tool for exploring data.

- Run a cell with **Shift+Enter**
- The kernel is the Python process that remembers variables from cell to cell
- Save often (File → Save, or the disk icon)

```python
print('Hello World!')
```

## 3. Variables persist across cells

Values created in one cell can be used in later cells, as long as the kernel is still running and you ran the earlier cell first.


```python
greeting = 'Hello'
recipient = 'Tim'
print(greeting + ', ' + recipient + '!')
```

## 4. Defining and calling a function

```python
def say_hello(recipient):
    """Return a greeting for the given recipient."""
    return 'Hello, {}!'.format(recipient)

say_hello('Tim')
```

## 6. Using numpy in a notebook

```python
import numpy as np

def square(x):
    return x * x

x = np.random.randint(1, 10)
y = square(x)
print('%d squared is %d' % (x, y))
```

## 7. The kernel remembers state

This cell reuses `x` and `square` from above. If it fails with `NameError`, run the previous cells first, or use **Kernel → Restart & Run All**.


```python
print('x is still', x)
print('square(x) is', square(x))
```

## 8. Plots with matplotlib

The `%matplotlib inline` magic tells Jupyter to show charts under the cell that created them.


```python
%matplotlib inline
import matplotlib.pyplot as plt

xs = [1, 2, 3, 4, 5]
ys = [1, 4, 9, 16, 25]
plt.plot(xs, ys, marker='o')
plt.title('y = x squared')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## 9. Magic commands

Magics start with `%` (line magic) or `%%` (cell magic). `%%timeit` measures how long a cell takes to run.


```python
import numpy as np
%timeit np.random.randint(1, 10)
```

```python
%%timeit
total = 0
for i in range(1000):
    total += i
```
